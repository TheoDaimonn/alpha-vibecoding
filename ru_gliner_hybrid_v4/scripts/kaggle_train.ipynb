```python
from pathlib import Path
import runpy
import shutil


input_root = Path("/kaggle/input")
mounted_projects = [
    path for path in input_root.iterdir()
    if path.is_dir() and (path / "scripts/kaggle_train.py").is_file()
]
if not mounted_projects:
    mounted = sorted(str(path) for path in input_root.iterdir()) if input_root.exists() else []
    raise FileNotFoundError(
        "Training bundle is not mounted. Expected a Kaggle Dataset containing "
        f"scripts/kaggle_train.py; mounted inputs: {mounted}"
    )

dataset_root = mounted_projects[0]
project_root = Path("/kaggle/working/project")
shutil.copytree(dataset_root, project_root, dirs_exist_ok=True)
runpy.run_path(str(project_root / "scripts/kaggle_train.py"), run_name="__main__")
```
